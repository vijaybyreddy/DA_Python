```python
# List Creation
age_list = [24, 25, 26, 27, 28]
name_list = ["Anand", "Bala", "Charan", "Divya", "Ezhil"]

#  List Operations
name_list.append("Yazhini")

# Insert 30 at index 2 in age_list
age_list.insert(2, 30)

# Remove "Yazhini" from name_list
name_list.remove("Yazhini")

#Pop the last element from age_list
age_list.pop()

# Extend age_list with additional ages [29, 30, 26]
age_list.extend([29, 30, 26])

# Sort age_list in descending order
age_list.sort(reverse=True)

# Find Max age, Min age, and sum of all ages
max_age = max(age_list)
min_age = min(age_list)
sum_ages = sum(age_list)

print("Updated Age List:", age_list)
print("Max Age:", max_age)
print("Min Age:", min_age)
print("Sum of Ages:", sum_ages)

# Accessing List Elements
# Print first element of name_list
print("First element:", name_list[0])

# Print last element of name_list
print("Last element:", name_list[-1])

#Print elements from index 2 to index 4
print("Index 2 to 4:", name_list[2:5])

# Print elements in reverse order
print("Reversed list:", name_list[::-1])
```

```python
# Prompt user for input score between 0 and 10
score = float(input("Enter your score (0 to 10): "))

if 0 <= score <= 10:
    if score > 7:
        print("Above Average: Outstanding work! Keep up the brilliant performance.")
    elif 4 <= score <= 7:
        print("Average: Good effort! Keep practicing, there's room for improvement.")
    else:
        print("Below Average: Need to improve your performance, consistent practice will lead to better results.")
else:
    print("Invalid score. Please enter a value between 0 and 10.")
```
